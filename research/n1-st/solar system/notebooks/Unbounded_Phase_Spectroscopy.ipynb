```python
import numpy as np
import matplotlib.pyplot as plt

print("Initializing Unbounded Phase Spectroscopy Pipeline...")
```

```python
def extract_unbounded_phase_spectrum(phi_field, phi_dot_field, center, size=64, n_bins=500000):
    """
    Extracts continuous Kuramoto phase spectrum from an unclipped Alpha Core subvolume.
    """
    z, y, x = center
    core_phi = phi_field[z-size:z+size, y-size:y+size, x-size:x+size]
    core_phi_dot = phi_dot_field[z-size:z+size, y-size:y+size, x-size:x+size]
    
    # Calculate Kuramoto phase angle
    core_theta = np.arctan2(core_phi_dot, core_phi)
    
    # Calculate energy density
    grad_z, grad_y, grad_x = np.gradient(core_phi)
    grad_sq = grad_x**2 + grad_y**2 + grad_z**2
    core_rho = 0.5 * (core_phi_dot**2 + grad_sq + core_phi**2)
    
    # Unbounded mass-phase histogram
    mass_spectrum, phase_bins = np.histogram(
        core_theta.flatten(), bins=n_bins, weights=core_rho.flatten()
    )
    return mass_spectrum, phase_bins
```
