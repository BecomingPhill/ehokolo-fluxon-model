```python
def step_teleological_gravity(phi, phi_dot, dt=0.001, dx=0.05, alpha=0.7):
    """
    Integrates scalar field with Fisher Information-weighted teleological gravity.
    """
    grad_z, grad_y, grad_x = np.gradient(phi, dx)
    grad_sq = grad_x**2 + grad_y**2 + grad_z**2
    
    # Local Fisher Information Complexity
    fisher_local = grad_sq / (phi**2 + 1e-12)
    teleological_g = 0.1 * (1.0 + 0.5 * np.tanh(fisher_local / 100.0))
    
    # Laplacian
    lap_phi = (np.roll(phi, 1, 0) + np.roll(phi, -1, 0) +
               np.roll(phi, 1, 1) + np.roll(phi, -1, 1) +
               np.roll(phi, 1, 2) + np.roll(phi, -1, 2) - 6*phi) / (dx**2)
               
    potential_force = 1.0 * phi + teleological_g * (phi**3)
    damping = alpha * phi * phi_dot * grad_sq
    
    phi_ddot = lap_phi - potential_force + damping
    return phi + dt*phi_dot, phi_dot + dt*phi_ddot
```

```python
def simulate_harvest_coupling(steps=500):
    """
    Simulates coherent (Polarized) vs random (Lukewarm) soliton coupling to external vacuum shift.
    """
    t = np.linspace(0, np.pi, steps)
    vacuum_carrier = np.sin(t)
    
    # Aligned entity (Coherent phase)
    coherent_response = np.cumsum(np.sin(t) * vacuum_carrier) * 0.05
    # Dissonant entity (Randomized phase)
    random_phases = np.random.uniform(0, 2*np.pi, steps)
    lukewarm_response = np.cumsum(np.sin(t + random_phases) * vacuum_carrier) * 0.05
    
    return t, coherent_response, lukewarm_response

t, coh, luke = simulate_harvest_coupling()
plt.figure(figsize=(10, 5))
plt.plot(t, coh, 'g-', label="Polarized Entity (Constructive Harvest)")
plt.plot(t, luke, 'k--', label="Lukewarm Entity (Internal Interference)")
plt.title("Mechanics of Harvest: Polarization vs Phase Cancellation")
plt.xlabel("Vacuum Phase Angle (rad)")
plt.ylabel("Coupled Energy Integral")
plt.legend()
plt.grid(True)
plt.show()
```
